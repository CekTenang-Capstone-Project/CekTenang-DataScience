# 04 - Dashboard Data Validation

Notebook validasi dashboard akan dibuat setelah processed data siap.

```python
import pandas as pd
```

```python
daily = pd.read_csv("../../../data/processed/daily_activities_clean.csv")
eda = pd.read_csv("../../../data/processed/daily_eda_dataset.csv")
predictions = pd.read_csv("../../../data/processed/stress_predictions_clean.csv")
weekly = pd.read_csv("../../../data/processed/weekly_summaries_clean.csv")
recommendations = pd.read_csv("../../../data/processed/recommendations_clean.csv")

print("Semua file berhasil di-load!")
```

```python
print("daily_activities_clean   :", daily.shape)
print("daily_eda_dataset        :", eda.shape)
print("stress_predictions_clean :", predictions.shape)
print("weekly_summaries_clean   :", weekly.shape)
print("recommendations_clean    :", recommendations.shape)
```

```python
print("=== daily_activities_clean ===")
print(daily.columns.tolist())

print("\n=== daily_eda_dataset ===")
print(eda.columns.tolist())

print("\n=== stress_predictions_clean ===")
print(predictions.columns.tolist())

print("\n=== weekly_summaries_clean ===")
print(weekly.columns.tolist())

print("\n=== recommendations_clean ===")
print(recommendations.columns.tolist())
```

```python
print("=== daily_activities_clean ===")
print(daily.isnull().sum())

print("\n=== daily_eda_dataset ===")
print(eda.isnull().sum())

print("\n=== stress_predictions_clean ===")
print(predictions.isnull().sum())

print("\n=== weekly_summaries_clean ===")
print(weekly.isnull().sum())

print("\n=== recommendations_clean ===")
print(recommendations.isnull().sum())
```

```python
print("stress_level (eda):", eda['stress_level'].unique())
print("stress_level (predictions):", predictions['stress_level'].unique())
print("dominant_stress_level (weekly):", weekly['dominant_stress_level'].unique())
print("stress_trend (weekly):", weekly['stress_trend'].unique())
print("main_trigger (weekly):", weekly['main_trigger'].unique())
print("category (recommendations):", recommendations['category'].unique())
print("priority_level (recommendations):", recommendations['priority_level'].unique())
print("period_type (recommendations):", recommendations['period_type'].unique())
```

```python
# Konversi kolom tanggal
daily['activity_date'] = pd.to_datetime(daily['activity_date'])
eda['activity_date'] = pd.to_datetime(eda['activity_date'])
predictions['prediction_date'] = pd.to_datetime(predictions['prediction_date'])
weekly['week_start'] = pd.to_datetime(weekly['week_start'])
weekly['week_end'] = pd.to_datetime(weekly['week_end'])

print("Konversi tanggal berhasil!")
print("Rentang tanggal daily     :", daily['activity_date'].min(), "s/d", daily['activity_date'].max())
print("Rentang tanggal weekly    :", weekly['week_start'].min(), "s/d", weekly['week_end'].max())
print("Rentang tanggal prediction:", predictions['prediction_date'].min(), "s/d", predictions['prediction_date'].max())
```

```python
print("=== HASIL VALIDASI DATA ===\n")

files = {
    "daily_activities_clean": daily,
    "daily_eda_dataset": eda,
    "stress_predictions_clean": predictions,
    "weekly_summaries_clean": weekly,
    "recommendations_clean": recommendations
}

for nama, df in files.items():
    total_missing = df.isnull().sum().sum()
    print(f"{nama}")
    print(f"  Rows    : {df.shape[0]}")
    print(f"  Columns : {df.shape[1]}")
    print(f"  Missing : {total_missing}")
    print(f"  Status  : {'OK' if total_missing == 0 else 'Ada missing values (cek catatan)'}")
    print()

print("Validasi selesai. Data siap digunakan untuk dashboard.")
```
